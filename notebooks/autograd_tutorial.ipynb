The steps of neural network training
- Forward Propagation (forward prop): NN takes best guess about correct output. Runs input data through functions to make a guess.

- Backward Propagation (backprop): NN adjusts paramaters proportionate to the error in its guess. Traverses backwards from the output, collectes error derivatives (gradients), and optimizes parameters using gradient descent.

```python
import torch
from torchvision.models import resnet18, ResNet18_Weights
model = resnet18(weights=ResNet18_Weights.DEFAULT)
data = torch.rand(1,3,64,64) #input data
labels = torch.rand(1,1000) #target output
```

```python
prediction = model(data) # forward pass
```

```python
loss = (prediction - labels).sum()
loss.backward() # backward pass
```

```python
#optimizer
optim = torch.optim.SGD(model.parameters(), lr=1e-2, momentum=0.9)
```

```python
optim.step()
```

```python
#requires_grad flag includes/excludes tensors from training
#set False to freeze parameters
x = torch.rand(5, 5)
y = torch.rand(5, 5)
z = torch.rand((5, 5), requires_grad=True)

a = x + y
print(f"Does `a` require gradients?: {a.requires_grad}")
b = x + z
print(f"Does `b` require gradients?: {b.requires_grad}")
```

```python
from torch import nn, optim

model = resnet18(weights=ResNet18_Weights.DEFAULT)

# Freeze all the parameters in the network
for param in model.parameters():
    param.requires_grad = False
```

```python
model.fc = nn.Linear(512, 10)
```

```python
# Optimize only the classifier
optimizer = optim.SGD(model.parameters(), lr=1e-2, momentum=0.9)
```
